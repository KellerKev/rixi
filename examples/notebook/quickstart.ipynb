# RIXI from a notebook

Drive a remote RIXI server with the importable SDK — no CLI, no shell alias. This is the
workflow the `rixi` PyPI package unlocks for data scientists and ML engineers.

**Prerequisites:** `pip install rixi` and a reachable RIXI server (locally:
`rixi-server --port 9000`, or an SSH tunnel to a GPU box).

```python
from rixi import Client

rixi = Client("http://127.0.0.1:9000")  # add token="…" if the server has auth enabled
rixi.health()
```

## Run a task and collect the result

`run()` packages the given directory, ships it, runs the Pixi task, and blocks until it
finishes — returning a `RunResult` with the collected output.

```python
result = rixi.run("../hello", task="hello")
print("ok:", result.ok)
print(result.output)
```

## Stream a long job live

For training or serving jobs, `stream()` yields output as it arrives so you can watch
progress in the notebook. Point it at the QLoRA example against a GPU box:

```python
for line in rixi.stream("../finetune-qlora", task="finetune"):
    print(line, end="")
```
